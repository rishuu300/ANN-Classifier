```python
import os
import pickle
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import load_model  # type: ignore
```

```python
# Load the model
model = tf.keras.models.load_model('model.h5')

# Load label encoder
with open('label_encoder.pkl', 'rb') as file:
    label_encoder = pickle.load(file)

# Load one-hot encoder
with open('onehot_encoder.pkl', 'rb') as file:
    onehot_encoder = pickle.load(file)

# Load scaler
with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
# Input Data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
input_df = pd.DataFrame([input_data])
```

```python
# Encode categorical vairables

# Label Encode 'Gender'
input_df['Gender'] = label_encoder.transform(input_df['Gender'])
```

```python
onehot_encoder.get_feature_names_out()
```

```python
# One-Hot-Encode 'Geography'
onehot_encoded = onehot_encoder.transform(input_df[['Geography']])
geo_encoded = pd.DataFrame(onehot_encoded, columns = onehot_encoder.get_feature_names_out())
```

```python
# Concatenate to original data
input_df = pd.concat([input_df.drop('Geography', axis = 1), geo_encoded], axis = 1)
```

```python
input_df
```

```python
# Scaling the data
input_scaled = scaler.transform(input_df)
```

```python
input_scaled
```

```python
prediction = model.predict(input_scaled)
```

```python
prediction
```

```python
if prediction[0][0] > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
