```python
import os
import pickle
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
```

```python
# Load the dataset
data_dir = 'D:\\Udemy\\Gen AI\\Projects\\ANN Classification\\data'
data = pd.read_csv(os.path.join(data_dir, 'Churn_Modelling.csv'))
data.head()
```

```python
# Preprocess the data
# Drop irrelevant columns

data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
data
```

```python
# Encode categorical variables
label_encoder = LabelEncoder()
data['Gender'] = label_encoder.fit_transform(data['Gender'])
```

```python
# One Hot Encoding for Geography
data['Geography'].unique()

from sklearn.preprocessing import OneHotEncoder
onehot_encoder = OneHotEncoder(sparse_output=False)

geo_encoder = onehot_encoder.fit_transform(data[['Geography']])
```

```python
geo_encoded = pd.DataFrame(geo_encoder, columns = onehot_encoder.get_feature_names_out(['Geography']))
```
